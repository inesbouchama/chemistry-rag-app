```python
import os
from PIL import Image
from IPython.display import display, Markdown
import glob

# === Configuration ===
SPEAKER = "Suleymanov (2)"  # Change to your speaker name
ROOT_DIR = "data/4m_data/english/test"
RGB_DIR = os.path.join(ROOT_DIR, "rgb", SPEAKER)

# === Get all image paths (jpg, jpeg, png) ===

image_paths = sorted(glob.glob(os.path.join(RGB_DIR, "*.[jJpP][pPnN][gG]")))

print("RGB_DIR:", RGB_DIR)
print("Found", len(image_paths), "image files")
print("First 3:", image_paths[:3])

# === Display slides ===
for img_path in image_paths:
    display(Markdown(f"### `{img_path}`"))
    display(Image.open(img_path).resize((800, 600)))  # Resize as needed
```
