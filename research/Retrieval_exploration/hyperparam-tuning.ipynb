```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the summary CSV
df = pd.read_csv("benchmark/data/grid_search_summary.csv")

# Heatmap: Accuracy vs hybrid_weight and temperature (averaging over top_p)
pivot = df.groupby(["temperature", "hybrid_weight"])["accuracy"].mean().unstack()

plt.figure(figsize=(10, 6))
sns.heatmap(pivot, annot=True, cmap="Blues", fmt=".2f", cbar_kws={'label': 'Accuracy'})
plt.title("📊 Accuracy Heatmap\n(Avg over top_p)")
plt.xlabel("Hybrid Weight")
plt.ylabel("Temperature")
plt.tight_layout()
plt.savefig("benchmark/data/heatmap_accuracy_temp_vs_hybrid.png")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

for temp in sorted(df['temperature'].unique()):
    subset = df[df['temperature'] == temp].groupby("hybrid_weight")["accuracy"].mean()
    plt.plot(subset.index, subset.values, marker="o", label=f"T={temp}")

plt.title("📈 Accuracy vs Hybrid Weight (Grouped by Temperature)")
plt.xlabel("Hybrid Weight")
plt.ylabel("Accuracy")
plt.legend(title="Temperature")
plt.grid(True)
plt.tight_layout()
plt.savefig("./benchmark/lineplot_accuracy_vs_hybrid.png")
plt.show()
```
